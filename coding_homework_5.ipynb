```python
# Import necessary libraries

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import spikes data files as numpy arrays

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv'
spikes_left_trial1 = np.loadtxt (url, delimiter = ',')

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
spikes_left_trial2 = np.loadtxt (url, delimiter = ',')

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv'
spikes_right_trial1 = np.loadtxt (url, delimiter = ',')

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
spikes_right_trial2 = np.loadtxt (url, delimiter = ',')
```

```python
# Print number of spikes in each trial

print(len(spikes_left_trial1))
print(len(spikes_right_trial1))
print(len(spikes_left_trial2))
print(len(spikes_right_trial2))
```

```python
# Calculate firing rates (spikes per second)

fr_left_trial1 = len(spikes_left_trial1) / 30
fr_right_trial1 = len(spikes_right_trial1) / 30
fr_left_trial2 = len(spikes_left_trial2) / 30
fr_right_trial2 = len(spikes_right_trial2) / 30

print(fr_left_trial1)
print(fr_left_trial2)
print(fr_right_trial1)
print(fr_right_trial2)
```

The firing rate of the neuron is higher during left turn trials. This tells me that the neuron tends to be more active when the rat is preparing to turn left.

```python
# Plot spike trains
fig, ax = plt.subplots(1, 2, figsize=(10,4))

# Trial 1
ax[0].scatter(spikes_left_trial1, np.ones_like(spikes_left_trial1), label='Left', color='red')
ax[0].scatter(spikes_right_trial1, np.ones_like(spikes_right_trial1)*2, label='Right', color='blue')
ax[0].set_title('Trial 1')
ax[0].set_xlabel('Time (s)')
ax[0].set_ylabel('Trial Type')

# Trial 2
ax[1].scatter(spikes_left_trial2, np.ones_like(spikes_left_trial2), label='Left', color='red')
ax[1].scatter(spikes_right_trial2, np.ones_like(spikes_right_trial2)*2, label='Right', color='blue')
ax[1].set_title('Trial 2')
ax[1].set_xlabel('Time (s)')

plt.legend()
plt.show()
```

The neuron appears more bursty during both trials for the left and right since there are frequent bursts or firings with spaces in between. However, it does seem like the left trials were more bursty while the right trials showed a more regular pattern, due to the clumps of firings with bigger gap breaks in the left turn trials versus the right turn trials which had firings that were a bit more spread out.   

```python
# Calculate ISI intervals

isi_left_trial1 = np.diff(spikes_left_trial1)
isi_right_trial1 = np.diff(spikes_right_trial1)
isi_left_trial2 = np.diff(spikes_left_trial2)
isi_right_trial2 = np.diff(spikes_right_trial2)
```

```python
# Defines the bins for the histogram
bins = np.arange(0, 2, 0.01)

# Creates the 4 subplots
fig, ax = plt.subplots(2, 2, figsize=(10,8))

# Left Trial 1
ax[0,0].hist(isi_left_trial1, bins=bins)
ax[0,0].set_title('ISI Left Trial 1')
ax[0,0].set_xlabel('Time (s)')
ax[0,0].set_ylabel('Frequency')

# Right Trial 1
ax[0,1].hist(isi_right_trial1, bins=bins)
ax[0,1].set_title('ISI Right Trial 1')
ax[0,1].set_xlabel('Time (s)')
ax[0,1].set_ylabel('Frequency')

# Left Trial 2
ax[1,0].hist(isi_left_trial2, bins=bins)
ax[1,0].set_title('ISI Left Trial 2')
ax[1,0].set_xlabel('Time (s)')
ax[1,0].set_ylabel('Frequency')

# Right Trial 2
ax[1,1].hist(isi_right_trial2, bins=bins)
ax[1,1].set_title('ISI Right Trial 2')
ax[1,1].set_xlabel('Time (s)')
ax[1,1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

Based on the histograms it looks like this neuron fires in bursts and has a high burstiness during the beginning of each trial. However, it would appear that once again, during the left turn trials, there is a slightly higher burstiness since there are more short intervals.

```python
# Time bins
bins = np.arange(0, 31, 1)

# Histogram counts (spikes per bin)
fr_lt1, _ = np.histogram(spikes_left_trial1, bins)
fr_rt1, _ = np.histogram(spikes_right_trial1, bins)
fr_lt2, _ = np.histogram(spikes_left_trial2, bins)
fr_rt2, _ = np.histogram(spikes_right_trial2, bins)

# Plot
fig, ax = plt.subplots(1, 2, figsize=(10,4))

# Trial 1
ax[0].plot(fr_lt1, color='red', label='Left')
ax[0].plot(fr_rt1, color='blue', label='Right')
ax[0].set_title('Trial 1')
ax[0].set_xlabel('Time (s)')
ax[0].set_ylabel('Firing Rate')

# Trial 2
ax[1].plot(fr_lt2, color='red', label='Left')
ax[1].plot(fr_rt2, color='blue', label='Right')
ax[1].set_title('Trial 2')
ax[1].set_xlabel('Time (s)')

plt.legend()
plt.show()
```

During the 30 seconds, the firing rate differs between left and right turn trials at certain times, particularly in Trial 1. In Trial 1, the neuron shows higher firing rates before left turns (especially around the middle of the delay period in the 10-16 second range), suggesting possible predictive activity for left choices. In contrast, right turn activity is generally lower and more sporadic, with a notable spike only near the end of the interval.

In Trial 2, however, the pattern is less distinct. The firing rates for left and right turns overlap and look pretty much identical, indicating little to no clear difference in activity between conditions.

Overall, Trial 1 suggests some separation in firing rate between left and right turns at specific time points, while Trial 2 shows much less differentiation.
