```python
#Import relevant modules

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import files as numpy arrays

url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv'
acc_hits = np.loadtxt (url1, delimiter = ',')

url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv'
acc_fa = np.loadtxt (url2, delimiter = ',')
```

```python
# Print the shape of each array

print("acc_hits shape:", acc_hits.shape)
print("acc_fa shape:", acc_fa.shape)
```

```python
# Create time variable from 0 to 4 seconds

time = np.linspace(0, 4, acc_hits.shape[1])
```

```python
# Calculate sampling rate

sampling_rate = 8001 / 4
print("Sampling rate:", sampling_rate, "Hz")
```

```python
# Plot first trial of hits and false alarms

plt.figure()

# Hits
plt.subplot(2,1,1)
plt.plot(time, acc_hits[0])
plt.title("First Trial - Hits")
plt.xlabel("Time (s)")
plt.ylabel("LFP")

# False alarms
plt.subplot(2,1,2)
plt.plot(time, acc_fa[0])
plt.title("First Trial - False Alarms")
plt.xlabel("Time (s)")
plt.ylabel("LFP")

plt.tight_layout()
plt.show()
```

I'm going to guess that the dominant oscillation appears to be relatively low frequency and is somewhat in the range of 1–5 Hz.

The acc_hits trial appears to show a more pronounced low frequency fluctuation, especially with a large deflection around 2.5–3 seconds, whereas the false alarm trial shows more irregular and slightly noisier fluctuations without a clearly dominant rhythm.

Overall, I'm pretty sure that the hits trial appears more structured, while the false alarm trial appears more variable and less rhythmic (perhaps more noisy, even).


```python
# Compute ERP for hits
erp_hits = np.mean(acc_hits, axis=0)

# Plot ERP
plt.plot(time, erp_hits)

# Vertical line at stimulus (2 sec)
plt.axvline(x=2, color='black')

# Horizontal zero line
plt.axhline(y=0, color='black')

plt.title("ERP - Hits")
plt.xlabel("Time (s)")
plt.ylabel("LFP")

plt.show()
```


The ERP for hits shows clear, consistent changes in the LFP around the time of the screen touch (2 seconds). Right after the vertical line, there is a large negative deflection followed by a strong positive peak. This looks like a pattern to me, and thus suggests that neural activity in the ACC is time locked to the screen touch during correct responses (indicating consistent and coordinated neural activity associated with hits).

```python
# Compute ERP for false alarms
erp_fa = np.mean(acc_fa, axis=0)

# Plot both ERPs
plt.plot(time, erp_hits, label="Hits")
plt.plot(time, erp_fa, label="False Alarms")

# Lines at event and zero
plt.axvline(x=2, color='black')
plt.axhline(y=0, color='black')

plt.title("ERP Comparison")
plt.xlabel("Time (s)")
plt.ylabel("LFP")

plt.legend()
plt.show()
```

There are differences between the ERPs for hits and false alarms. For example, the ERP for hits with a strong negative deflection followed by a positive peak. In contrast, the ERP for false alarms remains much smaller in amplitude and does not show the same clear pattern around the time of the response. What this indicates to me is that correct responses may involve more consistent or coordinated neural activity in the ACC compared to incorrect responses.
